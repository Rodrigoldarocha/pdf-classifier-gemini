## 📄 3. Gerando PDFs fictícios

Vamos criar **8 PDFs de exemplo** com dados fictícios que simulam documentos reais de uma empresa.

Os tipos são:
- 🧾 **Nota Fiscal** — vendas de produtos
- 🔧 **Ordem de Serviço (O.S.)** — serviços técnicos
- 📊 **Relatório** — resumo mensal
- 📋 **Planilha** — controle de estoque

```python
from reportlab.lib.pagesizes import A4
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib.units import cm
from reportlab.lib import colors
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer, Table, TableStyle
from reportlab.lib.enums import TA_CENTER, TA_LEFT, TA_RIGHT

styles = getSampleStyleSheet()

# ─── Estilo customizado ────────────────────────────────────────────────────────
titulo_style  = ParagraphStyle('titulo',  fontSize=16, alignment=TA_CENTER, spaceAfter=6,  fontName='Helvetica-Bold')
subtit_style  = ParagraphStyle('subtit',  fontSize=11, alignment=TA_CENTER, spaceAfter=14, textColor=colors.grey)
label_style   = ParagraphStyle('label',   fontSize=9,  textColor=colors.grey, spaceAfter=2)
valor_style   = ParagraphStyle('valor',   fontSize=11, spaceAfter=8,  fontName='Helvetica-Bold')
normal_style  = ParagraphStyle('normal',  fontSize=10, spaceAfter=6)
rodape_style  = ParagraphStyle('rodape',  fontSize=8,  textColor=colors.grey, alignment=TA_CENTER)

def linha_divisoria():
    return Table([['']], colWidths=[17*cm],
                 style=[('LINEBELOW', (0,0), (-1,-1), 0.5, colors.lightgrey),
                        ('TOPPADDING', (0,0), (-1,-1), 0),
                        ('BOTTOMPADDING', (0,0), (-1,-1), 6)])

print('✅ Estilos de documento configurados!')
```

```python
# ─── 📄 NOTA FISCAL ────────────────────────────────────────────────────────────
def criar_nota_fiscal(numero, cliente, itens, total, caminho):
    doc = SimpleDocTemplate(caminho, pagesize=A4,
                            rightMargin=2*cm, leftMargin=2*cm,
                            topMargin=2*cm, bottomMargin=2*cm)
    story = []
    story.append(Paragraph(f'NOTA FISCAL ELETRONICA - NF-e', titulo_style))
    story.append(Paragraph(f'Numero: {numero}  |  Serie: 001', subtit_style))
    story.append(linha_divisoria())

    story.append(Paragraph('EMITENTE', label_style))
    story.append(Paragraph('Tech Solutions Ltda — CNPJ: 12.345.678/0001-90', valor_style))
    story.append(Paragraph('Rua das Inovacoes, 500 — Recife, PE — CEP 50000-000', normal_style))
    story.append(linha_divisoria())

    story.append(Paragraph('DESTINATARIO', label_style))
    story.append(Paragraph(cliente['nome'], valor_style))
    story.append(Paragraph(f'CNPJ: {cliente["cnpj"]}  |  {cliente["endereco"]}', normal_style))
    story.append(linha_divisoria())

    story.append(Paragraph('ITENS DA NOTA', label_style))
    story.append(Spacer(1, 4))
    dados_tabela = [['Descricao', 'Qtd', 'Unit. (R$)', 'Total (R$)']]
    for item in itens:
        dados_tabela.append([item['desc'], str(item['qtd']),
                              f"{item['unit']:.2f}", f"{item['qtd']*item['unit']:.2f}"])

    t = Table(dados_tabela, colWidths=[8*cm, 2*cm, 3.5*cm, 3.5*cm])
    t.setStyle(TableStyle([
        ('BACKGROUND',    (0,0), (-1,0), colors.HexColor('#1a1a2e')),
        ('TEXTCOLOR',     (0,0), (-1,0), colors.white),
        ('FONTNAME',      (0,0), (-1,0), 'Helvetica-Bold'),
        ('FONTSIZE',      (0,0), (-1,-1), 9),
        ('ALIGN',         (1,0), (-1,-1), 'CENTER'),
        ('ROWBACKGROUNDS',(0,1), (-1,-1), [colors.white, colors.HexColor('#f5f5f5')]),
        ('GRID',          (0,0), (-1,-1), 0.3, colors.lightgrey),
        ('TOPPADDING',    (0,0), (-1,-1), 5),
        ('BOTTOMPADDING', (0,0), (-1,-1), 5),
    ]))
    story.append(t)
    story.append(Spacer(1, 10))
    story.append(Paragraph(f'<b>VALOR TOTAL: R$ {total:.2f}</b>', ParagraphStyle('total', fontSize=12, alignment=TA_RIGHT, textColor=colors.HexColor('#1a1a2e'))))
    story.append(Spacer(1, 20))
    story.append(Paragraph('Documento gerado automaticamente — dados fictícios para fins de demonstracao.', rodape_style))
    doc.build(story)

print('✅ Funcao criar_nota_fiscal() pronta!')
```

```python
# ─── 🔧 ORDEM DE SERVICO ───────────────────────────────────────────────────────
def criar_ordem_servico(numero, cliente, servicos, tecnico, status, caminho):
    doc = SimpleDocTemplate(caminho, pagesize=A4,
                            rightMargin=2*cm, leftMargin=2*cm,
                            topMargin=2*cm, bottomMargin=2*cm)
    story = []
    story.append(Paragraph('ORDEM DE SERVICO', titulo_style))
    story.append(Paragraph(f'OS Numero: {numero}  |  Status: {status}', subtit_style))
    story.append(linha_divisoria())

    story.append(Paragraph('CLIENTE', label_style))
    story.append(Paragraph(cliente['nome'], valor_style))
    story.append(Paragraph(f'Contato: {cliente["telefone"]}  |  {cliente["email"]}', normal_style))
    story.append(linha_divisoria())

    story.append(Paragraph('TECNICO RESPONSAVEL', label_style))
    story.append(Paragraph(tecnico, valor_style))
    story.append(linha_divisoria())

    story.append(Paragraph('SERVICOS REALIZADOS', label_style))
    story.append(Spacer(1, 4))
    dados = [['Servico', 'Horas', 'Valor (R$)']]
    total = 0
    for s in servicos:
        dados.append([s['desc'], f"{s['horas']}h", f"{s['valor']:.2f}"])
        total += s['valor']

    t = Table(dados, colWidths=[10*cm, 3*cm, 4*cm])
    t.setStyle(TableStyle([
        ('BACKGROUND',    (0,0), (-1,0), colors.HexColor('#16213e')),
        ('TEXTCOLOR',     (0,0), (-1,0), colors.white),
        ('FONTNAME',      (0,0), (-1,0), 'Helvetica-Bold'),
        ('FONTSIZE',      (0,0), (-1,-1), 9),
        ('ALIGN',         (1,0), (-1,-1), 'CENTER'),
        ('ROWBACKGROUNDS',(0,1), (-1,-1), [colors.white, colors.HexColor('#f0f8ff')]),
        ('GRID',          (0,0), (-1,-1), 0.3, colors.lightgrey),
        ('TOPPADDING',    (0,0), (-1,-1), 5),
        ('BOTTOMPADDING', (0,0), (-1,-1), 5),
    ]))
    story.append(t)
    story.append(Spacer(1, 10))
    story.append(Paragraph(f'<b>TOTAL: R$ {total:.2f}</b>', ParagraphStyle('total', fontSize=12, alignment=TA_RIGHT, textColor=colors.HexColor('#16213e'))))
    story.append(Spacer(1, 20))
    story.append(Paragraph('Documento gerado automaticamente — dados fictícios para fins de demonstracao.', rodape_style))
    doc.build(story)

print('✅ Funcao criar_ordem_servico() pronta!')
```

```python
# ─── 📊 RELATORIO MENSAL ───────────────────────────────────────────────────────
def criar_relatorio(mes, ano, dados_resumo, caminho):
    doc = SimpleDocTemplate(caminho, pagesize=A4,
                            rightMargin=2*cm, leftMargin=2*cm,
                            topMargin=2*cm, bottomMargin=2*cm)
    story = []
    story.append(Paragraph(f'RELATORIO OPERACIONAL', titulo_style))
    story.append(Paragraph(f'Periodo: {mes}/{ano}  |  Tech Solutions Ltda', subtit_style))
    story.append(linha_divisoria())

    story.append(Paragraph('RESUMO EXECUTIVO', label_style))
    story.append(Paragraph(
        f'Este relatorio apresenta o desempenho operacional do mes de {mes}/{ano}. '
        f'Foram emitidas {dados_resumo["total_nf"]} Notas Fiscais, '
        f'{dados_resumo["total_os"]} Ordens de Servico foram abertas e '
        f'{dados_resumo["os_fechadas"]} foram concluidas no periodo.',
    normal_style))
    story.append(linha_divisoria())

    story.append(Paragraph('INDICADORES DO MES', label_style))
    story.append(Spacer(1, 4))
    kpis = [
        ['Indicador', 'Valor'],
        ['Total de Notas Fiscais emitidas',  str(dados_resumo['total_nf'])],
        ['Faturamento bruto (R$)',            f"{dados_resumo['faturamento']:.2f}"],
        ['Ordens de Servico abertas',         str(dados_resumo['total_os'])],
        ['Ordens de Servico concluidas',      str(dados_resumo['os_fechadas'])],
        ['Ticket medio O.S. (R$)',            f"{dados_resumo['ticket_medio']:.2f}"],
        ['Clientes atendidos',                str(dados_resumo['clientes'])],
    ]
    t = Table(kpis, colWidths=[11*cm, 6*cm])
    t.setStyle(TableStyle([
        ('BACKGROUND',    (0,0), (-1,0), colors.HexColor('#0f3460')),
        ('TEXTCOLOR',     (0,0), (-1,0), colors.white),
        ('FONTNAME',      (0,0), (-1,0), 'Helvetica-Bold'),
        ('FONTSIZE',      (0,0), (-1,-1), 10),
        ('ALIGN',         (1,0), (-1,-1), 'CENTER'),
        ('ROWBACKGROUNDS',(0,1), (-1,-1), [colors.white, colors.HexColor('#e8f4fd')]),
        ('GRID',          (0,0), (-1,-1), 0.3, colors.lightgrey),
        ('TOPPADDING',    (0,0), (-1,-1), 6),
        ('BOTTOMPADDING', (0,0), (-1,-1), 6),
    ]))
    story.append(t)
    story.append(Spacer(1, 20))
    story.append(Paragraph('Documento gerado automaticamente — dados fictícios para fins de demonstracao.', rodape_style))
    doc.build(story)

print('✅ Funcao criar_relatorio() pronta!')
```

```python
# ─── 📋 PLANILHA / CONTROLE DE ESTOQUE ────────────────────────────────────────
def criar_planilha_estoque(referencia, itens, caminho):
    doc = SimpleDocTemplate(caminho, pagesize=A4,
                            rightMargin=2*cm, leftMargin=2*cm,
                            topMargin=2*cm, bottomMargin=2*cm)
    story = []
    story.append(Paragraph('CONTROLE DE ESTOQUE', titulo_style))
    story.append(Paragraph(f'Referencia: {referencia}  |  Tech Solutions Ltda', subtit_style))
    story.append(linha_divisoria())

    story.append(Paragraph('ITENS EM ESTOQUE', label_style))
    story.append(Spacer(1, 4))
    dados = [['Codigo', 'Produto', 'Qtd Atual', 'Qtd Minima', 'Status']]
    for item in itens:
        status = 'OK' if item['qtd'] >= item['minimo'] else 'REPOR'
        dados.append([item['codigo'], item['produto'], str(item['qtd']), str(item['minimo']), status])

    t = Table(dados, colWidths=[2.5*cm, 7*cm, 2.5*cm, 2.5*cm, 2.5*cm])
    t.setStyle(TableStyle([
        ('BACKGROUND',    (0,0), (-1,0), colors.HexColor('#533483')),
        ('TEXTCOLOR',     (0,0), (-1,0), colors.white),
        ('FONTNAME',      (0,0), (-1,0), 'Helvetica-Bold'),
        ('FONTSIZE',      (0,0), (-1,-1), 9),
        ('ALIGN',         (0,0), (-1,-1), 'CENTER'),
        ('ROWBACKGROUNDS',(0,1), (-1,-1), [colors.white, colors.HexColor('#f3eeff')]),
        ('GRID',          (0,0), (-1,-1), 0.3, colors.lightgrey),
        ('TOPPADDING',    (0,0), (-1,-1), 5),
        ('BOTTOMPADDING', (0,0), (-1,-1), 5),
    ]))
    story.append(t)
    story.append(Spacer(1, 20))
    story.append(Paragraph('Documento gerado automaticamente — dados fictícios para fins de demonstracao.', rodape_style))
    doc.build(story)

print('✅ Funcao criar_planilha_estoque() pronta!')
```

## 🗂️ 5. Coletando e catalogando os PDFs (simulando o `collector.py`)

```python
from pathlib import Path
import json
from datetime import datetime

def coletar_pdfs(pasta_entrada: str) -> list[dict]:
    """
    Percorre a pasta de entrada e retorna uma lista com
    metadados de cada PDF encontrado.
    Simula o comportamento do collector.py em producao.
    """
    pasta = Path(pasta_entrada)
    pdfs_encontrados = []

    for pdf in sorted(pasta.glob('*.pdf')):
        info = {
            'arquivo':        pdf.name,
            'caminho':        str(pdf),
            'tamanho_kb':     round(pdf.stat().st_size / 1024, 1),
            'coletado_em':    datetime.now().strftime('%Y-%m-%d %H:%M:%S'),
            'status':         'aguardando_extracao',
        }
        pdfs_encontrados.append(info)
        print(f'  📄 {pdf.name:45s} {info["tamanho_kb"]:6.1f} KB')

    return pdfs_encontrados


print('🔍 Coletando PDFs em pdfs_entrada/...\n')
catalogo = coletar_pdfs('pdfs_entrada')

# Salva o catálogo como JSON
with open('resultados/catalogo_pdfs.json', 'w', encoding='utf-8') as f:
    json.dump(catalogo, f, ensure_ascii=False, indent=2)

print(f'\n✅ {len(catalogo)} PDFs catalogados!')
print('💾 Catalogo salvo em: resultados/catalogo_pdfs.json')
```
